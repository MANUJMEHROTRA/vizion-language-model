# Vision-Language Models: study roadmap

A hands-on path from tensors to a small vision-language model (VLM), then to a safe SO-101 robot learning experiment. The notebooks are intentionally framework-light: use PyTorch, NumPy, pandas, and Python's standard library only. No Transformers, timm, Lightning, or other model/training wrappers are used. A tiny model trained here is an educational instrument, not a competitive pretrained foundation model.

## How to study

1. Read the concept, write down the tensor shapes, then run each code cell in order.
2. Change one variable at a time; record seed, data split, config, loss, and results in your experiment notes.
3. Do the exercises before reading later notebooks. Reproduce the toy results with at least three seeds.
4. Never connect an unvalidated policy to an energized robot. Begin in simulation/offline replay and keep a human-operated emergency stop.

## Prerequisites and setup

Know Python functions/classes, linear algebra, probability, gradient descent, and basic PyTorch tensors/autograd. Select a Python kernel with `torch`, `numpy`, and `pandas` installed. Use the official [PyTorch install selector](https://pytorch.org/get-started/locally/) for your OS and accelerator; install NumPy and pandas in that same kernel. Most examples are CPU-sized. The full from-scratch notebook uses a synthetic dataset so it runs without downloads or credentials.

## What is and is not included

- VLM: input includes image(s) and text, output is text. It does not by itself produce safe motor commands.
- VLA: conditions a policy on vision, language, and usually robot state to produce actions or action chunks.
- Pretraining from random initialization at useful scale requires curated data, substantial compute, infrastructure, and evaluations; this curriculum demonstrates mechanics, not that scale.
- Reported benchmark scores are not interchangeable. Compare the same dataset version, prompt, resolution, decoding, and evaluation protocol.

## Suggested capstone

Train notebook 04, freeze the image encoder and apply notebook 05's LoRA idea, then design an offline SO-101 evaluation protocol in notebook 08. Do not deploy until the policy passes replay, simulation, workspace-limited low-speed tests, and an emergency-stop drill.
