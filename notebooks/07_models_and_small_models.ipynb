# 07 — Model families, small models, and the 2026 landscape snapshot

**Checked 2026-10-17.** “Top VLM” has no universal answer: an OCR leader may not be best at charts, long video, spatial grounding, latency, or local deployment. Use the dynamic Open VLM leaderboard and the task-specific model card evaluation with a fixed protocol before making a claim. The candidates below are a shortlist for evaluation, not an independently validated ordinal ranking. Vendor-reported benchmark numbers are self-reported and protocols differ.

## General multimodal candidates to compare

| Family/checkpoint | Why it is on the shortlist | Good evaluation targets | Important caveat |
|---|---|---|---|
| Qwen3.5 (e.g. 4B/9B) | Current open-weight multimodal family; hybrid architecture and broad image/language capability claimed in its card | OCR, VQA, long-context, spatial and reasoning slices | Validate exact task scores and deployment/runtime support; card language includes vendor claims |
| Gemma 4 E2B/E4B, 12B, 26B A4B, 31B | Multimodal size range from edge-oriented to larger; E2B/E4B and 12B cards describe image plus audio; 26B is MoE | image QA, latency/quality trade-off, multilingual, device fit | “Effective” and total parameters differ for small models; MoE total parameters exceed active parameters |

## Widely useful small language / multimodal models

| Model family | Type | Typical fit |
|---|---|---|
| Gemma 4 E2B/E4B | compact multimodal; image and audio input described in model card | edge/local visual assistant; test memory, latency, license and prompt fit |
| Qwen3.5 4B | compact multimodal | local image-text reasoning and robotics semantic front-end experiments |
| SmolVLM2 256M/500M/2.2B | compact image/video-to-text | resource-constrained visual QA and video summaries |
| Qwen3 0.6B/1.7B (text-only variants) | text SLM | instruction parsing, task labels, lightweight language component; not an image encoder |
| SmolLM2 135M/360M/1.7B (text-only) | text SLM | small text generation/classification and educational baselines |

SLM is a size/efficiency description, not a guarantee of reasoning, safety, or real-time performance. For SO-101, do not choose only by parameter count. Measure visual task success, action latency, memory, quantization effects, and reliable behavior under lighting/camera variation.

## Robotics models are a different category: VLA

| Work/model | What it contributes | Where it fits |
|---|---|---|
| OpenVLA (7B) | Open VLA trained on 970k robot demonstrations; paper reports a mixture of DINOv2/SigLIP vision and Llama 2 language components and parameter-efficient adaptation | baseline for action-conditioned transfer if hardware/compute and action representation fit |
| SmolVLA | Compact VLA designed for affordable robot data and consumer hardware; paper describes action chunks and flow matching | promising SO-101-style experimentation candidate; validate exact policy/repo version |
| Octo | Generalist robot policy trained on diverse robot data | compare multi-task generalization and target action/state interfaces |
| RT-2 / RT-X family | Landmark vision-language-action line; includes closed and research systems | important conceptually; not necessarily downloadable or suitable for local use |

A VLM can describe “the blue block is left of the cup”; it cannot safely convert that sentence into calibrated servo trajectories without a policy/controller, state, action representation, and safety layer. VLA behavior is trained on robot trajectories, not just captions.

## Benchmark reading checklist

Before believing a rank: inspect benchmark version, split, contamination controls, system prompt, image resolution/tiling, test-time tools, number of attempts, answer normalization, reasoning mode, and whether score was independently reproduced. For robot papers, inspect embodiment, control frequency, action chunk size, number of demonstrations, success definition, and real-world versus simulation conditions.
